# Raw .shp polygons

```python
import geopandas as gpd
import pandas as pd
import warnings

warnings.filterwarnings("ignore", category=UserWarning)

soil_gdf = gpd.read_file("D:\PersonalProjects\pagsubay-datasci\soil-type\Philippine-Soil-Series\Philippine Soil Series (shapefile)\PhilSoilSeries.shp")
mdf = pd.read_csv ("D:\PersonalProjects\pagsubay-datasci\soil-type\phivolcs_mData.csv")
```

```python
eq_gdf = gpd.GeoDataFrame(
    mdf, 
    geometry=gpd.points_from_xy(mdf['longitude'], mdf['latitude']),
    crs="EPSG:4326"
)

# 2. Force CRS alignment
if soil_gdf.crs is None or soil_gdf.crs.to_epsg() != 4326:
    soil_gdf = soil_gdf.to_crs("EPSG:4326")

# 3. Run the spatial join
matched_data = gpd.sjoin(eq_gdf, soil_gdf, how="left", predicate="within")
matched_data = matched_data[~matched_data.index.duplicated(keep='first')]
```

```python
# 4. Weight assignment function
def assign_soil_weight(description):
    desc = str(description).lower()
    if 'rock' in desc or 'volcanic' in desc or 'lava' in desc or 'mountain' in desc:
        return 1.3  
    elif 'loam' in desc or 'dense' in desc:
        return 1.05 
    elif 'sand' in desc or 'alluvial' in desc:
        return 0.8  
    elif 'clay' in desc or 'swamp' in desc or 'delta' in desc:
        return 0.5  
    elif 'hydrosol' in desc or 'reclaimed' in desc:
        return 0.3  
    return 1.0 

# 5. Extract using your actual schema (event_id, longitude, latitude)
final_output = matched_data[['event_id', 'longitude', 'latitude', 'SOILDESC']].copy()
final_output['Site_Weight'] = final_output['SOILDESC'].apply(assign_soil_weight)

# Save to CSV
final_output.to_csv(r"D:\PersonalProjects\pagsubay-datasci\soil-type\Earthquake_Soil_Weights.csv", index=False)
print(final_output.head())
```
