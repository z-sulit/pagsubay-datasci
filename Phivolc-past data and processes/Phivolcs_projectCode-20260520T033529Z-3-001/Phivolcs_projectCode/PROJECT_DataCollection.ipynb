##Extraction (October 2025)

```python
#fetching on that specific link (For october)
#iba naman sa september and onwards ha

url = "https://earthquake.phivolcs.dost.gov.ph"
response = requests.get(url, verify=False) #skipping verification gang

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    print(soup.prettify()[:500])
    print("Success!")
else:
    print("Failed fetcher")
```

```python
all_rows = []
headers = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
           "Depth (km)", "Mag", "Location"]

tables = soup.find_all("table", {"class": "MsoNormalTable"})

for table in tables:
    ths = [th.get_text(strip=True) for th in table.find_all("th")]
    if ths and "Date - Time" in ths[0]:
        for tr in table.find_all("tr")[1:]:
            cols = [td.get_text(strip=True) for td in tr.find_all("td")]
            if cols:
                all_rows.append(cols)

if all_rows:
    df = pd.DataFrame(all_rows, columns=headers)

    #split sesame
    df[["Coordinates", "Location"]] = df["Location"].str.split("of ", n=1, expand=True)
    df["Coordinates"] = df["Coordinates"].str.strip()
#-------ocd lang to
    cols = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
        "Depth (km)", "Mag", "Coordinates", "Location"]
    df = df[cols]
#-------ocd lang to
    df.to_csv("phivolcs_earthquakes(October_2025).csv", index=False, encoding="utf-8")
    print(f"Data Saved: {len(df)} rows")
    display(df.head())
else:
    print("No Data")
```

```python
pages = {
    "2024/2024_January": "January_2024",
    "2024/2024_February": "February_2024",
    "2024/2024_March": "March_2024",
    "2024/2024_April": "April_2024",
    "2024/2024_May": "May_2024",
    "2024/2024_June": "June_2024",
    "2024/2024_July": "July_2024",
    "2024/2024_August": "August_2024",
    "2024/2024_September": "September_2024",
    "2024/2024_October": "October_2024",
    "2024/2024_November": "November_2024",
    "2024/2024_December": "December_2024",
}

headers = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
           "Depth (km)", "Mag", "Location"]

for path, filename in pages.items():
    url = f"https://earthquake.phivolcs.dost.gov.ph/EQLatest-Monthly/{path}.html"
    print(f"\nFetching: {url}")
    response = requests.get(url, verify=False)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, "html.parser")
        print("Success!")
    else:
        print("Failed fetcher")
        continue

    all_rows = []
    tables = soup.find_all("table", {"class": "MsoNormalTable"})

    for table in tables:
        ths = [th.get_text(strip=True) for th in table.find_all("th")]
        if ths and "Date - Time" in ths[0]:
            for tr in table.find_all("tr")[1:]:
                cols = [td.get_text(strip=True) for td in tr.find_all("td")]
                if cols:
                    all_rows.append(cols)

    if all_rows:
        df = pd.DataFrame([row[:6] for row in all_rows], columns=headers)

        df[["Coordinates", "Location"]] = df["Location"].str.split("of ", n=1, expand=True)
        df["Coordinates"] = df["Coordinates"].str.strip() + " of"

        cols = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
                "Depth (km)", "Mag", "Coordinates", "Location"]
        df = df[cols]

        out_file = f"phivolcs_earthquakes({filename}).csv"
        df.to_csv(out_file, index=False, encoding="utf-8")
        print(f"Data Saved: {len(df)} rows to {out_file}")

        cnt = df.duplicated().sum()
        duplicates = df[df.duplicated(keep=False)]
        print("Duplicate rows:", cnt)
        if cnt > 0:
            print(duplicates.head())

        print("\nNull counts:")
        print(df.isnull().sum())

    else:
        print("No Data")
```
