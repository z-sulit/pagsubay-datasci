```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import urllib3
```

#checking and cleaning october (2025)

```python
url = "https://earthquake.phivolcs.dost.gov.ph/"
response = requests.get(url, verify=False) #skipping verification gang

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    print(soup.prettify()[:500])
    print("Success!")
else:
    print("Failed fetcher")
```

```python
all_rows = []
headers = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
           "Depth (km)", "Mag", "Location"]

tables = soup.find_all("table", {"class": "MsoNormalTable"})

for table in tables:
    ths = [th.get_text(strip=True) for th in table.find_all("th")]
    if ths and "Date - Time" in ths[0]:
        for tr in table.find_all("tr")[1:]:
            cols = [td.get_text(strip=True) for td in tr.find_all("td")]
            if cols:
                all_rows.append(cols)

if all_rows:

    df = pd.DataFrame([row[:6] for row in all_rows], columns=headers)

    #split sesame
    df[["Coordinates", "Location"]] = df["Location"].str.split("of ", n=1, expand=True)
    df["Coordinates"] = df["Coordinates"].str.strip()

#-----ocd lang to
    cols = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
            "Depth (km)", "Mag", "Coordinates", "Location"]
    df = df[cols]


    #df.to_csv("phivolcs_earthquakes(july test).csv", index=False, encoding="utf-8")
    #print(f"Data Saved: {len(df)} rows")
    display(df.head())

else:
    print("No Data")
```

```python
cnt = df.duplicated().sum()

duplicates = df[df.duplicated(keep=False)]
print(duplicates)
print("\n")
print("Dups No:", cnt)
print("\n")
print("This is an earthquake data so duplicates are okay. Just make sure to double check the website and confirm.")
```

```python
df = df.drop_duplicates(keep="first")
```

```python
cnt = df.duplicated().sum()

duplicates = df[df.duplicated(keep=False)]
print(duplicates)
print("\n")
print("Dups No:", cnt)
print("\n")
print("This is an earthquake data so duplicates are okay. Just make sure to double check the website and confirm.")
```

```python
    df.to_csv("phivolcs_earthquakes_2025_10_October.csv", index=False, encoding="utf-8")
    print(f"Data Saved: {len(df)} rows")
    display(df.head())
```

#wtf is happening (2018) Jan - May

##January

```python
import re
import pandas as pd

all_rows = []
headers = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
           "Depth (km)", "Mag", "Location"]


page_text = soup.get_text(" ", strip=True)
CURRENT_YEAR = 2018
CURRENT_MONTH = "January"

tables = soup.find_all("table", {"class": "MsoNormalTable"})

for table in tables:
    rows = table.find_all("tr")

    header_index = None
    for i, tr in enumerate(rows):
        cells = [cell.get_text(strip=True) for cell in tr.find_all(["th","td","p","span"])]
        if cells and "Date" in cells[0]:
            header_index = i
            break

    if header_index is not None:
        for tr in rows[header_index+1:]:
            tds = tr.find_all("td")
            if not tds:
                continue

            cols = []
            for j, td in enumerate(tds):
                if j == 0:
                    # Prefer full text in <a>
                    link = td.find("a")
                    if link and re.search(r"\d{1,2}\s+\w+\s+\d{4}", link.get_text(strip=True)):
                        dt = link.get_text(strip=True)
                    else:

                        pieces = [s for s in td.stripped_strings]
                        joined = " ".join(pieces)


                        if re.fullmatch(r"\d{1,2}", joined) and len(pieces) > 1:
                            day = joined

                            m_time = re.search(r"(\d{1,2}:\d{2}\s*(AM|PM))", " ".join(pieces[1:]), flags=re.I)
                            if m_time:
                                dt = f"{day} {CURRENT_MONTH} {CURRENT_YEAR} - {m_time.group(1)}"
                            else:
                                dt = f"{day} {CURRENT_MONTH} {CURRENT_YEAR}"
                        else:
                            dt = joined
                    cols.append(dt)
                else:

                    cols.append(td.get_text(strip=True))

            if cols and len(cols) >= 6:
                all_rows.append(cols[:6])
        break

if all_rows:
    df = pd.DataFrame(all_rows, columns=headers)


    parts = df["Location"].str.split(r"\s*of\s*", n=1, regex=True, expand=True)
    df["Coordinates"] = parts[0].str.strip()
    df["Location"]    = parts[1].fillna("").str.strip()

    df.loc[df["Location"] == "", "Location"] = df["Coordinates"]
    df.loc[df["Location"] == df["Coordinates"], "Coordinates"] = None


    df = df.replace("Â°", "°", regex=True)


    cols = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
            "Depth (km)", "Mag", "Coordinates", "Location"]
    df = df[cols]

    df.to_csv("phivolcs_earthquakes_2018_03_March.csv", index=False, encoding="utf-8-sig")
    print(f"Data Saved: {len(df)} rows")
    display(df.head())
else:
    print("No Data")
```

```python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
import urllib3

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

MONTHS_FULL = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

HEADERS = ["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
           "Depth (km)", "Mag", "Location"]

def scrape_one_month(year: int, month_full: str) -> pd.DataFrame:
    url = f"https://earthquake.phivolcs.dost.gov.ph/EQLatest-Monthly/{year}/{year}_{month_full}.html"
    print(f"\n📂 Fetching {url} …")

    try:
        r = requests.get(url, verify=False, timeout=30)
    except Exception as e:
        print(f"  ⚠️ error: {e}")
        return pd.DataFrame(columns=HEADERS + ["Coordinates"])

    if r.status_code != 200:
        print(f"  ⚠️ {month_full} {year}: page not found ({r.status_code})")
        return pd.DataFrame(columns=HEADERS + ["Coordinates"])

    soup = BeautifulSoup(r.text, "html.parser")
    tables = soup.find_all("table", {"class": "MsoNormalTable"})
    if not tables:
        print("  ⚠️ no tables found")
        return pd.DataFrame(columns=HEADERS + ["Coordinates"])

    all_rows = []
    for table in tables:
        rows = table.find_all("tr")

        # Identify header row by keyword
        header_index = None
        for i, tr in enumerate(rows):
            txt = " ".join([c.get_text(" ", strip=True).lower() for c in tr.find_all(["th","td","p","span"])])
            if "date" in txt and "location" in txt:
                header_index = i
                break

        if header_index is None:
            continue

        for tr in rows[header_index+1:]:
            tds = tr.find_all("td")
            if not tds or len(tds) < 6:
                continue

            # ✅ Date-time (prefer <a> if present)
            date_td = tds[0]
            link = date_td.find("a")
            if link:
                dt = link.get_text(strip=True)
            else:
                dt = date_td.get_text(" ", strip=True)

            if not dt or dt.upper() in MONTHS_FULL or re.fullmatch(r"\d{1,2}", dt):
                continue

            # ✅ The rest of the row
            rest = [td.get_text(" ", strip=True) for td in tds[1:6]]

            row = [dt] + rest
            if len(row) == 6:
                all_rows.append(row)

        if all_rows:
            break

    if not all_rows:
        print("  ⚠️ no rows")
        return pd.DataFrame(columns=HEADERS + ["Coordinates"])

    df = pd.DataFrame(all_rows, columns=HEADERS)

    # Split "Coordinates" and "Location"
    parts = df["Location"].str.split(r"\s*of\s*", n=1, regex=True, expand=True)
    df["Coordinates"] = parts[0].str.strip()
    df["Location"]    = parts[1].fillna("").str.strip()
    df.loc[df["Location"] == "", "Location"] = df["Coordinates"]
    df.loc[df["Location"] == df["Coordinates"], "Coordinates"] = None

    df = df.replace("Â°", "°", regex=True)

    # Reorder
    df = df[["Date - Time (Philippine Time)", "Latitude (ºN)", "Longitude (ºE)",
             "Depth (km)", "Mag", "Coordinates", "Location"]]

    # Save CSV
    month_idx = MONTHS_FULL.index(month_full) + 1
    out_name = f"phivolcs_earthquakes_{year}_{month_idx:02d}_{month_full}.csv"
    df.to_csv(out_name, index=False, encoding="utf-8-sig")

    # Duplicates
    duplicates = df[df.duplicated()]
    cnt = len(duplicates)
    if cnt > 0:
        print("Duplicates found:\n", duplicates)
    print("\nDups No:", cnt)
    print("This is earthquake data so duplicates are okay. Double check with website if unsure.")

    # Nulls
    print("\nNulls per column:\n", df.isnull().sum())

    print(f"  ✅ saved {len(df)} rows → {out_name}")
    return df


def scrape_year(year: int):
    for m in MONTHS_FULL:
        scrape_one_month(year, m)


# -------- RUN --------
scrape_year(2017)
```
